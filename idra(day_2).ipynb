```python
import random


def get_valid_guess(low, high):
    """Prompt the user until a valid integer guess within range is entered."""
    while True:
        raw = input(f"Enter your guess ({low}-{high}): ").strip()
        if not raw.lstrip("-").isdigit():
            print("Please enter a whole number.")
            continue

        guess = int(raw)
        if guess < low or guess > high:
            print(f"Your guess must be between {low} and {high}.")
            continue

        return guess


def play_game(low=1, high=100, max_attempts=7):
    secret_number = random.randint(low, high)
    attempts_used = 0

    print("=" * 40)
    print("       WELCOME TO THE GUESSING GAME")
    print("=" * 40)
    print(f"I'm thinking of a number between {low} and {high}.")
    print(f"You have {max_attempts} attempts to guess it. Good luck!\n")

    while attempts_used < max_attempts:
        attempts_left = max_attempts - attempts_used
        print(f"Attempts remaining: {attempts_left}")

        guess = get_valid_guess(low, high)
        attempts_used += 1

        if guess == secret_number:
            print(f"\nCorrect! You guessed it in {attempts_used} attempt(s).")
            print("Congratulations, you win!")
            return
        elif guess < secret_number:
            print("Too low. Try again.\n")
        else:
            print("Too high. Try again.\n")

    print(f"\nOut of attempts! The number was {secret_number}.")
    print("Better luck next time!")


def play_again():
    while True:
        choice = input("Play again? (y/n): ").strip().lower()
        if choice in ("y", "yes"):
            return True
        if choice in ("n", "no"):
            return False
        print("Please answer 'y' or 'n'.")


def main():
    while True:
        play_game(low=1, high=100, max_attempts=7)
        print()
        if not play_again():
            print("Thanks for playing. Goodbye!")
            break
        print()


if __name__ == "__main__":
    main()
```
