```python
import pandas as pd
import numpy as np
df = pd.read_csv('Day8_Ecommerce_Sales_Dataset.csv')
df.head()
```

```python

df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df.info()
```

```python
print("Shape (rows, columns):", df.shape)
print("\nColumns:", list(df.columns))
print("\nMissing values per column:\n", df.isnull().sum())
df.select_dtypes(include='number').describe()
```

## 2. Data Selection

Selecting specific columns and rows using label-based (`.loc`) and position-based (`.iloc`) indexing.

```python

sales_view = df[['Order_ID', 'Product', 'Category', 'City', 'Payment_Method', 'Total_Sales']]
sales_view.head()
```

```python

df.iloc[0:5]
```

```python

df.loc[df['Order_ID'] == 1010, ['Order_ID', 'Customer', 'Product', 'Total_Sales', 'Payment_Method']]
```

```python

high_value_orders = df[df['Total_Sales'] > 10000]
print(f"Number of high-value orders (> ₹10,000): {len(high_value_orders)}")
high_value_orders[['Order_ID', 'Product', 'City', 'Total_Sales']]
```

```python

electronics_upi = df[(df['Category'] == 'Electronics') & (df['Payment_Method'] == 'UPI')]
electronics_upi[['Order_ID', 'Product', 'City', 'Total_Sales', 'Payment_Method']]
```

```python

discounted_top_rated = df[(df['Discount_Percent'] > 0) & (df['Rating'] >= 4.5)]
print(f"Discounted orders rated 4.5+: {len(discounted_top_rated)}")
discounted_top_rated[['Order_ID', 'Product', 'Discount_Percent', 'Rating', 'Total_Sales']].head()
```

```python

city_orders = df[df['City'] == 'Hyderabad']
city_orders[['Order_ID', 'Product', 'Category', 'Total_Sales']]
```

```python

top5_orders = df.sort_values('Total_Sales', ascending=False).head(5)
top5_orders[['Order_ID', 'Product', 'City', 'Total_Sales']]
```

```python

bottom5_orders = df.sort_values('Total_Sales', ascending=True).head(5)
bottom5_orders[['Order_ID', 'Product', 'City', 'Total_Sales']]
```

```python

multi_sorted = df.sort_values(['Category', 'Total_Sales'], ascending=[True, False])
multi_sorted[['Category', 'Order_ID', 'Product', 'Total_Sales']].head(10)
```

```python
category_summary = df.groupby('Category').agg(
    total_sales=('Total_Sales', 'sum'),
    avg_sales=('Total_Sales', 'mean'),
    max_sales=('Total_Sales', 'max'),
    min_sales=('Total_Sales', 'min'),
    total_quantity=('Quantity', 'sum'),
    num_orders=('Order_ID', 'count')
).round(2).sort_values('total_sales', ascending=False)

category_summary
```

```python
city_summary = df.groupby('City').agg(
    total_sales=('Total_Sales', 'sum'),
    avg_sales=('Total_Sales', 'mean'),
    total_quantity=('Quantity', 'sum'),
    num_orders=('Order_ID', 'count')
).round(2).sort_values('total_sales', ascending=False)

city_summary
```

```python

city_summary.head(5)
```

```python
product_summary = df.groupby('Product').agg(
    total_sales=('Total_Sales', 'sum'),
    total_quantity=('Quantity', 'sum'),
    num_orders=('Order_ID', 'count'),
    avg_rating=('Rating', 'mean')
).round(2).sort_values('total_sales', ascending=False)

print("Top 5 best-selling products by total sales:")
product_summary.head(5)
```

```python

top_qty_product = product_summary.sort_values('total_quantity', ascending=False).head(1)
top_qty_product
```

```python
payment_summary = df.groupby('Payment_Method').agg(
    total_sales=('Total_Sales', 'sum'),
    avg_sales=('Total_Sales', 'mean'),
    num_orders=('Order_ID', 'count')
).round(2).sort_values('total_sales', ascending=False)

payment_summary
```

```python
category_payment = df.groupby(['Category', 'Payment_Method'])['Total_Sales'].agg(['sum', 'count']).round(2)
category_payment.rename(columns={'sum': 'total_sales', 'count': 'num_orders'})
```

```python
overall_stats = {
    'Total Sales (all orders)': df['Total_Sales'].sum(),
    'Average Sales per Order': df['Total_Sales'].mean(),
    'Highest Single Order Value': df['Total_Sales'].max(),
    'Lowest Single Order Value': df['Total_Sales'].min(),
    'Total Quantity Sold': df['Quantity'].sum(),
    'Total Number of Orders': df['Order_ID'].nunique()
}

for k, v in overall_stats.items():
    print(f"{k}: {round(v, 2) if isinstance(v, float) else v}")
```

```python

print("Highest-value order:")
display(df.loc[df['Total_Sales'].idxmax()])

print("\nLowest-value order:")
display(df.loc[df['Total_Sales'].idxmin()])
```

```python
top_category = category_summary['total_sales'].idxmax()
top_product = product_summary['total_sales'].idxmax()
top_city = city_summary['total_sales'].idxmax()
top_payment = payment_summary['total_sales'].idxmax()

print(f"Top-performing category : {top_category} (₹{category_summary.loc[top_category, 'total_sales']:,.2f})")
print(f"Top-performing product  : {top_product} (₹{product_summary.loc[top_product, 'total_sales']:,.2f})")
print(f"Top city by sales       : {top_city} (₹{city_summary.loc[top_city, 'total_sales']:,.2f})")
print(f"Most-used payment method: {top_payment} (₹{payment_summary.loc[top_payment, 'total_sales']:,.2f})")
```
