```python

import pandas as pd
import numpy as np
from google.colab import files


try:
    customers = pd.read_csv('Day9_Customers.csv')
    products = pd.read_csv('Day9_Products.csv')
    orders = pd.read_csv('Day9_Orders.csv')
    print("Files loaded successfully from Colab directory!")
except FileNotFoundError:
    print("Please upload Day9_Customers.csv, Day9_Products.csv, and Day9_Orders.csv:")
    uploaded = files.upload()
    customers = pd.read_csv('Day9_Customers.csv')
    products = pd.read_csv('Day9_Products.csv')
    orders = pd.read_csv('Day9_Orders.csv')


orders_customers = pd.merge(orders, customers, on='Customer_ID', how='left')

merged_df = pd.merge(orders_customers, products, on='Product_ID', how='left')


north_west_orders = merged_df[merged_df['Region'].isin(['North', 'West'])].copy()
south_east_orders = merged_df[merged_df['Region'].isin(['South', 'East'])].copy()


combined_df = pd.concat([north_west_orders, south_east_orders], ignore_index=True)


combined_df['Total_Amount'] = combined_df.apply(
    lambda row: row['Quantity'] * row['Unit_Price'], axis=1
)


def calculate_discount(row):
    membership = row['Membership_Type']
    total_amount = row['Total_Amount']

    if membership == 'Premium':
        return total_amount * 0.10
    elif membership == 'Regular':
        return total_amount * 0.05
    else:
        return 0.0

combined_df['Discount_Amount'] = combined_df.apply(calculate_discount, axis=1)

combined_df['Final_Amount'] = combined_df['Total_Amount'] - combined_df['Discount_Amount']


combined_df['Order_Date'] = pd.to_datetime(combined_df['Order_Date'])

combined_df['Year'] = combined_df['Order_Date'].dt.year
combined_df['Month'] = combined_df['Order_Date'].dt.month_name()
combined_df['Day'] = combined_df['Order_Date'].dt.day
combined_df['Day_of_Week'] = combined_df['Order_Date'].dt.day_name()
combined_df['Quarter'] = combined_df['Order_Date'].dt.quarter


ordered_columns = [
    'Order_ID', 'Order_Date', 'Year', 'Month', 'Day', 'Day_of_Week', 'Quarter',
    'Customer_ID', 'Customer_Name', 'City', 'Region', 'Membership_Type',
    'Product_ID', 'Product_Name', 'Category', 'Brand', 'Unit_Price', 'Quantity',
    'Total_Amount', 'Discount_Amount', 'Final_Amount', 'Payment_Method', 'Order_Status'
]

final_df = combined_df[ordered_columns].sort_values(by='Order_ID').reset_index(drop=True)

output_filename = 'Processed_Ecommerce_Dataset.csv'
final_df.to_csv(output_filename, index=False)

print("\n--- DATASET PROCESSING SUMMARY ---")
print(f"Total Rows Processed: {final_df.shape[0]}")
print(f"Total Columns Created: {final_df.shape[1]}")
print("\nFirst 5 rows of processed dataset:")
display(final_df.head())


files.download(output_filename)
```
