```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv")
```

```python
df = df.drop_duplicates()
```

```python
str_cols = [
    "Employee_Name",
    "Department",
    "Job_Title",
    "Gender",
    "City",
    "Work_Mode",
]
for col in str_cols:
  df[col] = df[col].astype(str).str.strip().str.title()
  df[col] = df[col].replace({"Nan": np.nan, "None": np.nan})
```

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
```

```python
for col in ["Department", "Gender", "City", "Work_Mode"]:
  df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median()).astype(int)
df["Experience_Years"] = (
    df["Experience_Years"].fillna(df["Experience_Years"].median()).round(1)
)
```

```python
df["Annual_Salary"] = df.groupby("Department")["Annual_Salary"].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df["Performance_Score"] = (
    df["Performance_Score"].fillna(df["Performance_Score"].mode()[0]).astype(int)
)
```

```python
df.to_csv("Cleaned_Company_Employee_Dataset.csv", index=False)
```
